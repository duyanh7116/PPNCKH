```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.metrics import accuracy_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Bidirectional, LSTM, Dense, Flatten

# 1. Hàm tiền xử lý dữ liệu chuẩn pFedCross
def preprocess_data(data):
    # Xử lý giá trị thiếu bằng trung vị (median)[cite: 1]
    data = data.fillna(data.median(numeric_only=True))

    X = data.iloc[:, :-1].values
    y = data.iloc[:, -1].values

    # Chuẩn hóa Min-Max [0,1] và mã hóa nhãn[cite: 1]
    scaler = MinMaxScaler()
    X_scaled = scaler.fit_transform(X)

    encoder = LabelEncoder()
    y_encoded = encoder.fit_transform(y)

    return X_scaled, y_encoded

# 2. Khởi tạo mô hình DCNN-BiLSTM[cite: 1]
def create_dcnn_bilstm(input_shape, num_classes):
    model = Sequential([
        Conv1D(filters=64, kernel_size=3, activation='relu', input_shape=input_shape),
        MaxPooling1D(pool_size=2),
        Bidirectional(LSTM(64, return_sequences=False)),
        Dense(32, activation='relu'),
        Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Khởi tạo 2 thuật toán cơ bản khác để đủ yêu cầu 3 thuật toán
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

# 3. Kịch bản thực nghiệm (3 tập dữ liệu x 3 thuật toán = 9 kết quả)
datasets = {
    "CICIoT2023_Subset1": (np.random.rand(1000, 20), np.random.randint(0, 5, 1000)), # Dữ liệu giả lập thay cho CICIoT2023[cite: 1]
    "Dataset_2_Kaggle": (np.random.rand(1000, 20), np.random.randint(0, 5, 1000)),
    "Dataset_3_UCI": (np.random.rand(1000, 20), np.random.randint(0, 5, 1000))
}

algorithms = {
    "DCNN-BiLSTM": None, # Sẽ khởi tạo động theo shape
    "Random Forest": RandomForestClassifier(),
    "Decision Tree": DecisionTreeClassifier()
}

results = []

for data_name, (X_raw, y_raw) in datasets.items():
    # Giả lập chia train/test đơn giản
    X_train, y_train = X_raw[:800], y_raw[:800]
    X_test, y_test = X_raw[800:], y_raw[800:]

    for algo_name, model in algorithms.items():
        if algo_name == "DCNN-BiLSTM":
            # Reshape cho DCNN
            X_train_dl = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
            X_test_dl = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

            dl_model = create_dcnn_bilstm((X_train_dl.shape[1], 1), len(np.unique(y_train)))
            dl_model.fit(X_train_dl, y_train, epochs=1, verbose=0)

            _, acc = dl_model.evaluate(X_test_dl, y_test, verbose=0)
            results.append({"Dataset": data_name, "Algorithm": algo_name, "Accuracy": acc})
        else:
            model.fit(X_train, y_train)
            preds = model.predict(X_test)
            acc = accuracy_score(y_test, preds)
            results.append({"Dataset": data_name, "Algorithm": algo_name, "Accuracy": acc})

# 4. Biểu diễn kết quả (Biểu đồ)
df_results = pd.DataFrame(results)
pivot_df = df_results.pivot(index='Dataset', columns='Algorithm', values='Accuracy')

pivot_df.plot(kind='bar', figsize=(10, 6))
plt.title('So sánh độ chính xác: 3 Thuật toán x 3 Tập dữ liệu')
plt.ylabel('Độ chính xác (Accuracy)')
plt.xticks(rotation=0)
plt.legend(title='Thuật toán')
plt.tight_layout()
plt.show()
```
